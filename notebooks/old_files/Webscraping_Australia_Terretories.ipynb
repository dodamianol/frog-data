```python
import pandas as pd
```

```python
columns_to_load_territories_states = ['Postal', 'Population (Sept 2023)[17]', 'Area (km2)[18]']
url = "https://en.wikipedia.org/wiki/States_and_territories_of_Australia"

States_and_territories_of_Australia = pd.read_html(url, header=0)

## ---------------------------------------------------------
## wahrscheinlich sollte ich hier "https://en.wikipedia.org/wiki/States_and_territories_of_Australia"
## einfach die tabelle nehmen die bei statistics ist
```

```python
# getting data on australias states and cleaning it
states_of_australia = States_and_territories_of_Australia[2]
states_of_australia_filtered = states_of_australia[['State'] + columns_to_load_territories_states]
states_of_australia_filtered = states_of_australia_filtered.rename(columns={'State': 'State or Territory'})

# getting data on australias terretories and cleaning it + erasing 'Jervis Bay Territory'
territories_of_australia = States_and_territories_of_Australia[3]
territories_of_australia = territories_of_australia[territories_of_australia['Territory'] != 'Jervis Bay Territory']
territories_of_australia_filtered = territories_of_australia[['Territory'] + columns_to_load_territories_states]
territories_of_australia_filtered = territories_of_australia_filtered.rename(columns={'Territory': 'State or Territory'})
```

```python
states_and_territories = pd.merge(states_of_australia_filtered, territories_of_australia_filtered, how='outer').set_index('Postal')
states_and_territories = states_and_territories.rename(columns={'Population (Sept 2023)[10]': 'Population (Sept 2023)'})
states_and_territories = states_and_territories.rename(columns={'Area (km2)[11]': 'Area (km2)'})

states_and_territories
```
