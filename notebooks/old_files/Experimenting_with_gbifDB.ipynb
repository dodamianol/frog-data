```python
import pandas as pd
```

```python
filename = "data/FrogID4_gbif_database.csv"

# load the relevant columns only
# (others are either empty, contail all the same information or contain irrelevant information)
columns_to_load = [ 
      'gbifID', 'family', 'genus', 'species', 'stateProvince',
      'decimalLatitude', 'decimalLongitude', 'coordinateUncertaintyInMeters',
      'eventDate', 'day', 'month', 'year',
      'taxonKey', 'recordedBy' 
]
```

```python
frogID_database = pd.read_csv(filename, usecols=columns_to_load, sep='\t', on_bad_lines='skip')
frogID_database
```

```python
recordedBy_counts = frogID_database.groupby('recordedBy').size().reset_index(name='counts')
recordedBy_counts = recordedBy_counts.sort_values(by='counts', ascending=False)
recordedBy_counts
```
